```python
import random
import ipywidgets as widgets
from IPython.display import display, HTML

# =========================================================
# 🍓 FRUIT HANGMAN GAME
# =========================================================

fruits = [
    "banana",
    "cherry",
    "orange",
    "papaya",
    "grapes"
]

word = ""
guessed_letters = []
wrong_guesses = 0
max_wrong_guesses = 6
game_running = False


# =========================================================
# 🎨 DASHBOARD STYLE
# =========================================================

display(HTML("""
<style>

.fruit-dashboard {
    background: linear-gradient(135deg, #063b24, #087f45, #0aa85f);
    padding: 30px;
    border-radius: 25px;
    color: white;
    text-align: center;
    font-family: Arial, sans-serif;
    box-shadow: 0px 8px 25px rgba(0,0,0,0.25);
    margin-bottom: 15px;
}

.game-title {
    font-size: 32px;
    font-weight: bold;
    margin-bottom: 8px;
}

.game-subtitle {
    font-size: 16px;
    color: #d8ffe9;
}

.card {
    background: white;
    color: #063b24;
    padding: 20px;
    border-radius: 18px;
    margin-top: 15px;
    font-size: 18px;
    box-shadow: 0px 4px 15px rgba(0,0,0,0.15);
}

.word-card {
    background: #f4fff8;
    border: 3px solid #0aa85f;
    color: #063b24;
    padding: 25px;
    border-radius: 20px;
    text-align: center;
    font-size: 32px;
    font-weight: bold;
    letter-spacing: 8px;
}

.status-card {
    background: #ffffff;
    color: #063b24;
    padding: 18px;
    border-radius: 15px;
    font-size: 17px;
    margin-top: 15px;
}

</style>
"""))


# =========================================================
# 🏠 HEADER
# =========================================================

header = widgets.HTML("""
<div class="fruit-dashboard">
    <div class="game-title">🍓 FRUIT HANGMAN 🍓</div>

    <div class="game-subtitle">
        Guess the hidden 6-letter fruit!
    </div>

    <div style="font-size:35px; margin-top:15px;">
    
    </div>

    <div style="margin-top:15px;">
        🎯 6 Chances &nbsp; | &nbsp; 🍓 5 Fruits
    </div>
</div>
""")

display(header)


# =========================================================
# 🎮 BUTTONS
# =========================================================

start_button = widgets.Button(
    description="▶ START GAME",
    button_style="success",
    layout=widgets.Layout(
        width="170px",
        height="45px"
    )
)

new_button = widgets.Button(
    description="🔄 NEW GAME",
    button_style="warning",
    layout=widgets.Layout(
        width="170px",
        height="45px"
    )
)

end_button = widgets.Button(
    description="■ END GAME",
    button_style="danger",
    layout=widgets.Layout(
        width="170px",
        height="45px"
    )
)

button_box = widgets.HBox(
    [start_button, new_button, end_button],
    layout=widgets.Layout(
        justify_content="center",
        gap="12px",
        margin="15px 0"
    )
)

display(button_box)


# =========================================================
# 🔤 INPUT
# =========================================================

guess_input = widgets.Text(
    placeholder="Enter one letter...",
    description="Letter:",
    disabled=True,
    layout=widgets.Layout(width="300px")
)

guess_button = widgets.Button(
    description="GUESS",
    button_style="info",
    disabled=True,
    layout=widgets.Layout(
        width="120px",
        height="35px"
    )
)

input_box = widgets.HBox(
    [guess_input, guess_button],
    layout=widgets.Layout(
        justify_content="center",
        margin="10px 0"
    )
)

display(input_box)


# =========================================================
# 📊 GAME OUTPUT
# =========================================================

game_output = widgets.Output()

display(game_output)


# =========================================================
# 📊 UPDATE DASHBOARD
# =========================================================

def update_dashboard():

    with game_output:

        game_output.clear_output(wait=True)

        if not game_running:

            display(HTML("""
            <div class="card" style="text-align:center;">
                🎮 Click <b>START GAME</b> to begin!
            </div>
            """))

            return

        # Hidden word
        display_word = ""

        for letter in word:

            if letter in guessed_letters:
                display_word += letter.upper() + " "
            else:
                display_word += "_ "

        # Remaining chances
        remaining = max_wrong_guesses - wrong_guesses

        hearts = "❤️ " * remaining

        if remaining == 0:
            hearts = "💔"

        # Word display
        display(HTML(f"""
        <div class="word-card">
            {display_word}
        </div>
        """))

        # Guessed letters
        if guessed_letters:

            guessed_text = ", ".join(
                letter.upper()
                for letter in guessed_letters
            )

        else:

            guessed_text = "None"

        # Status
        display(HTML(f"""
        <div class="status-card">

            <b>❌ Wrong Guesses:</b>
            {wrong_guesses} / {max_wrong_guesses}

            &nbsp;&nbsp;&nbsp;

            <b>❤️ Chances:</b>
            {hearts}

            <br><br>

            <b>🔤 Guessed Letters:</b>
            {guessed_text}

        </div>
        """))


# =========================================================
# ▶ START GAME
# =========================================================

def start_game(button):

    global word
    global guessed_letters
    global wrong_guesses
    global game_running

    word = random.choice(fruits)

    guessed_letters = []

    wrong_guesses = 0

    game_running = True

    guess_input.disabled = False
    guess_button.disabled = False

    guess_input.value = ""

    update_dashboard()


# =========================================================
# 🔄 NEW GAME
# =========================================================

def new_game(button):

    global word
    global guessed_letters
    global wrong_guesses
    global game_running

    # Select completely new random fruit
    word = random.choice(fruits)

    guessed_letters = []

    wrong_guesses = 0

    game_running = True

    # Enable game controls
    guess_input.disabled = False
    guess_button.disabled = False

    guess_input.value = ""

    with game_output:

        game_output.clear_output(wait=True)

        display(HTML("""
        <div class="card"
             style="text-align:center;
                    border:3px solid #f0ad4e;">

            <h2>🔄 NEW GAME STARTED!</h2>

            <p>
                🎯 A new fruit has been selected.
            </p>

            <p>
                Good luck! 
            </p>

        </div>
        """))

    update_dashboard()


# =========================================================
# ■ END GAME
# =========================================================

def end_game(button):

    global game_running

    if game_running:

        game_running = False

        guess_input.disabled = True
        guess_button.disabled = True

        with game_output:

            game_output.clear_output(wait=True)

            display(HTML(f"""
            <div class="card"
                 style="text-align:center;
                        border:3px solid #dc3545;">

                <h1 style="color:#dc3545;">
                    🛑 GAME ENDED
                </h1>

                <p>
                    You ended the current game.
                </p>

                <p>
                    The fruit was:
                    <b>{word.upper()}</b>
                </p>

                <p>
                    🔄 Click <b>NEW GAME</b>
                    to start another game.
                </p>

            </div>
            """))


# =========================================================
# 🎯 GUESS LETTER
# =========================================================

def make_guess(button):

    global wrong_guesses
    global game_running

    if not game_running:
        return

    guess = guess_input.value.lower().strip()

    # Empty input
    if guess == "":

        with game_output:

            print("⚠️ Please enter a letter.")

        return

    # Invalid input
    if len(guess) != 1 or not guess.isalpha():

        with game_output:

            print("⚠️ Please enter ONE alphabet letter only.")

        guess_input.value = ""

        return

    # Duplicate letter
    if guess in guessed_letters:

        with game_output:

            print("🔄 You already guessed this letter!")

        guess_input.value = ""

        return

    # Save guess
    guessed_letters.append(guess)

    # Correct guess
    if guess in word:

        print("✅ Correct guess! ")

    # Wrong guess
    else:

        wrong_guesses += 1

        print("❌ Wrong guess!")

    guess_input.value = ""

    # =====================================================
    # 🏆 WIN
    # =====================================================

    if all(letter in guessed_letters for letter in word):

        game_running = False

        guess_input.disabled = True
        guess_button.disabled = True

        with game_output:

            game_output.clear_output(wait=True)

            display(HTML(f"""
            <div class="card"
                 style="text-align:center;
                        border:3px solid #0aa85f;">

                <h1 style="color:#087f45;">
                    🎉 YOU WIN! 🎉
                </h1>

                <h2>
                     {word.upper()}
                </h2>

                <p>
                    Congratulations!
                    You guessed the fruit correctly!
                </p>

                <p>
                    🏆 Excellent Work!
                </p>

                <p>
                    🔄 Click <b>NEW GAME</b>
                    to play again.
                </p>

            </div>
            """))

        return

    # =====================================================
    # 💀 GAME OVER
    # =====================================================

    if wrong_guesses >= max_wrong_guesses:

        game_running = False

        guess_input.disabled = True
        guess_button.disabled = True

        with game_output:

            game_output.clear_output(wait=True)

            display(HTML(f"""
            <div class="card"
                 style="text-align:center;
                        border:3px solid #dc3545;">

                <h1 style="color:#dc3545;">
                    💀 GAME OVER
                </h1>

                <h2>
                    {word.upper()}
                </h2>

                <p>
                    You used all 6 incorrect guesses.
                </p>

                <p>
                    🔄 Click <b>NEW GAME</b>
                    to try again!
                </p>

            </div>
            """))

        return

    update_dashboard()


# =========================================================
# 🔗 BUTTON CONNECTIONS
# =========================================================

start_button.on_click(start_game)

new_button.on_click(new_game)

end_button.on_click(end_game)

guess_button.on_click(make_guess)

guess_input.on_submit(make_guess)


# =========================================================
# 🚀 INITIAL SCREEN
# =========================================================

update_dashboard()
```
